````python
import time
import psutil
import torch
from transformers import (
    AutoTokenizer,
    AutoModel,
    AutoModelForCausalLM
)

# Set up device - 'cuda' for GPU, 'cpu' for fallback
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Text for verification - long enough to register computational load
TEST_TEXT = """
Machine learning techniques are increasingly used for phishing detection.
Transformers provide strong contextual understanding.
"""

# -------------------------
# Load Models
# -------------------------
bert_model_name = "bert-base-uncased"
gpt_model_name = "gpt2"

print(f"Loading Models onto device: {DEVICE.upper()}...")

# Load BERT Components (Encoder)
bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
bert_model = AutoModel.from_pretrained(bert_model_name).to(DEVICE)

# Load GPT Components (Decoder)
gpt_tokenizer = AutoTokenizer.from_pretrained(gpt_model_name)
if gpt_tokenizer.pad_token is None:
    gpt_tokenizer.pad_token = gpt_tokenizer.eos_token
gpt_model = AutoModelForCausalLM.from_pretrained(gpt_model_name).to(DEVICE)

# -------------------------
# Optimized Benchmark Function
# -------------------------
def benchmark(model, tokenizer, text, model_type):
    process = psutil.Process()
    
    # --- Structural Fix for Tokenizer Crash ---
    # Force text into a clean string format to ensure Step 1 success
    clean_text = str(text).strip()

    # Pre-tokenize outside the timing loop
    tokens = tokenizer(
        clean_text,
        return_tensors="pt",
        padding=True,
        truncation=True
    )
    tokens = {k: v.to(DEVICE) for k, v in tokens.items()}

    # --- ADDED: Warm-up Run ---
    # Forces weights/VRAM allocations before timing starts
    with torch.no_grad():
        if model_type == "BERT": _ = model(**tokens)
        else: _ = model.generate(tokens["input_ids"], max_new_tokens=2, pad_token_id=tokenizer.pad_token_id)

    # --- ADDED: CUDA Sync and Snapshot Memory ---
    if DEVICE == "cuda": torch.cuda.synchronize()
    memory_before = process.memory_info().rss / 1024**2
    
    start = time.time()
    with torch.no_grad():
        if model_type == "BERT":
            # Encoder path: Single forward pass over context
            output = model(**tokens)
            processed_tokens = tokens["input_ids"].shape[1]
        else:
            # Decoder path: Autoregressive loop generating NEW tokens
            output = model.generate(
                tokens["input_ids"],
                max_new_tokens=50,
                pad_token_id=tokenizer.pad_token_id
            )
            # Correct calculation: Only count NEWLY MINTED tokens
            processed_tokens = output.shape[1] - tokens["input_ids"].shape[1]

    # --- CRITICAL ADDITION: Synchronize GPU and Stopwatch ---
    # Wait until GPU math is done before stopping the stopwatch!
    if DEVICE == "cuda": torch.cuda.synchronize()
    end = time.time()
    
    memory_after = process.memory_info().rss / 1024**2
    latency = end - start

    # Safely handle potential negative memory delta on CPU
    memory_usage = max(0, memory_after - memory_before)

    return {
        "Latency(sec)": round(latency, 4),
        "Tokens": processed_tokens,
        "Memory(MB)": round(memory_usage, 2),
        "Tokens/sec": round(processed_tokens / latency, 2)
    }

# -------------------------
# Run Benchmarks
# -------------------------
print("Starting benchmarks...")
bert_results = benchmark(bert_model, bert_tokenizer, TEST_TEXT, "BERT")
gpt_results = benchmark(gpt_model, gpt_tokenizer, TEST_TEXT, "GPT")

# -------------------------
# Print Comparative Summary Table
# -------------------------
print(f"\n===== BENCHMARK Results ({DEVICE.upper()}) =====")

# Clean comparative table layout
print(f"{'Metric':<18} | {'BERT (Encoder)':<16} | {'GPT2 (Decoder)':<16}")
print("-" * 56)
for metric in bert_results.keys():
    print(f"{metric:<18} | {bert_results[metric]:<16} | {gpt_results[metric]:<16}")

print("\nArchitectural Context Specs:")
print(" -> BERT Context Window: 512 Tokens")
print(" -> GPT2 Context Window: 1024 Tokens")
````

````python
import time
import psutil
import os

# Filter non-critical warnings
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3" 
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)

import torch
# Safety Check 1: Ensure Flash-Attn is NOT present (source of the crash)
try:
    import flash_attn
    print("Warning: Flash Attention is still installed and may cause crashes.")
except ImportError:
    pass # This is the good path! Standard torch attention will be used.

from transformers import AutoTokenizer, AutoModel, AutoModelForCausalLM

# Set hardware acceleration context
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Verified Device Environment: {DEVICE.upper()}")

TEST_TEXT = """
Machine learning techniques are increasingly used for phishing detection.
Transformers provide strong contextual understanding.
"""

# Define local checkpoints
bert_model_name = "bert-base-uncased"
gpt_model_name = "gpt2"

print("\nInitializing Neural Networks into Memory...")
DualModelSuccess = True

try:
    # --- Load BERT Encoder Pipeline ---
    bert_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
    # bert-base-uncased doesn't rely on flash-attn
    bert_model = AutoModel.from_pretrained(bert_model_name).to(DEVICE)
    print(" -> BERT Encoder Pipeline: Ready")
except Exception as e:
    DualModelSuccess = False
    print(f" -> BERT Initialization Error: {str(e)}")

try:
    # --- Load GPT2 Decoder Pipeline ---
    # With flash-attn gone, GPT2 will safely use PyTorch's native attention engine.
    gpt_tokenizer = AutoTokenizer.from_pretrained(gpt_model_name)
    if gpt_tokenizer.pad_token is None:
        gpt_tokenizer.pad_token = gpt_tokenizer.eos_token
    gpt_model = AutoModelForCausalLM.from_pretrained(gpt_model_name).to(DEVICE)
    print(" -> GPT2 Decoder Pipeline: Ready")
except Exception as e:
    DualModelSuccess = False
    print(f" -> GPT2 Initialization Error (Attention Conflict Fix Failure): \n{str(e)}")

# -----------------------------------------------------
# BENCHMARK ENGINE (Encoder vs. Decoder profiling)
# -----------------------------------------------------
def benchmark(model, tokenizer, text, model_type):
    process = psutil.Process()
    clean_text = str(text).strip()

    # Pre-tokenize (Step 1 of modular inference)
    tokens = tokenizer(clean_text, return_tensors="pt", padding=True, truncation=True)
    tokens = {k: v.to(DEVICE) for k, v in tokens.items()}

    # Warm-up pass to populate system cache states and prevent cold-start lag
    with torch.no_grad():
        if model_type == "BERT": _ = model(**tokens)
        else: _ = model.generate(tokens["input_ids"], max_new_tokens=2, pad_token_id=tokenizer.pad_token_id)

    # Pre-math synchronization hook
    if DEVICE == "cuda": torch.cuda.synchronize()
    memory_before = process.memory_info().rss / 1024**2
    
    start = time.time()
    with torch.no_grad():
        if model_type == "BERT":
            # Path A (BERT): Parallel processing through a single forward pass
            output = model(**tokens)
            processed_tokens = tokens["input_ids"].shape[1]
        else:
            # Path B (GPT): Step-by-step Autoregressive Generation Loop
            output = model.generate(
                tokens["input_ids"],
                max_new_tokens=50,
                pad_token_id=tokenizer.pad_token_id
            )
            # Correct measurement counts only NEWLY produced tokens
            processed_tokens = output.shape[1] - tokens["input_ids"].shape[1]

    # Post-math synchronization hook (Required for accurate latency)
    if DEVICE == "cuda": torch.cuda.synchronize()
    end = time.time()
    
    memory_after = process.memory_info().rss / 1024**2
    latency = end - start

    # Return standard performance metrics dictionary
    return {
        "Latency(sec)": round(latency, 4),
        "Tokens": processed_tokens,
        "Memory(MB)": round(max(0, memory_after - memory_before), 2),
        "Tokens/sec": round(processed_tokens / latency, 2)
    }

# Execute Evaluation Run Loops
if DualModelSuccess:
    print("\nInitialization success! Executing model profiling benchmarks...")
    bert_results = benchmark(bert_model, bert_tokenizer, TEST_TEXT, "BERT")
    gpt_results = benchmark(gpt_model, gpt_tokenizer, TEST_TEXT, "GPT")

    print(f"\n===== BENCHMARK PROFILE RESULTS ({DEVICE.upper()}) =====")
    print(f"{'Performance Metric':<18} | {'BERT (Encoder)':<16} | {'GPT2 (Decoder)':<16}")
    print("-" * 56)
    for metric in bert_results.keys():
        print(f"{metric:<18} | {bert_results[metric]:<16} | {gpt_results[metric]:<16}")
else:
    print("\nUnable to run benchmarks due to model initialization failures.")
    print("Please paste the NEW initialization error messages.")
````
